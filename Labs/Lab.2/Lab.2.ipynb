*Exercise 1:* Write a function that creates an **n×n matrix** (a list of lists) representing the state of a tic‑tac‑toe game.

Use the integers:

- `0` = empty
- `1` = `"X"`
- `2` = `"O"`


````python
# Exercise 1: Create an n x n board
# 0 = empty, 1 = X, 2 = O
def create_board(n):
    board = []
    for i in range(n):
        row = []
        for j in range(n):
            row.append(0)
        board.append(row)
    return board
````

*Exercise 3:* Modify Exercise 2 so that it takes a matrix in the format from Exercise 1 and draws a tic‑tac‑toe board with `"X"`s and `"O"`s.

````python
# Exercise 3: Draw board showing X and O from the matrix
def draw_board_state(board):
    n = len(board)
    m = len(board[0])

    for i in range(n):
        # print top border of row
        print("  " + " ---" * m)

        # building the cell line for this row
        line = "  |"
        for j in range(m):
            if board[i][j] == 0:
                line = line + "   |"
            elif board[i][j] == 1:
                line = line + " X |"
            else:
                line = line + " O |"
        print(line)

    # print final bottom border
    print("  " + " ---" * m)
````

*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Exercise 4: Check the game status
# Returns: 1 if player 1 wins, 2 if player 2 wins, 0 if draw, -1 if still going
def check_winner(board):
    n = len(board)

    for player in [1, 2]:

        # Check each row
        for i in range(n):
            won = True
            for j in range(n):
                if board[i][j] != player:
                    won = False
            if won:
                return player

        # Check each column
        for j in range(n):
            won = True
            for i in range(n):
                if board[i][j] != player:
                    won = False
            if won:
                return player

        # Check main diagonal (top-left to bottom-right)
        won = True
        for i in range(n):
            if board[i][i] != player:
                won = False
        if won:
            return player

        # Check anti-diagonal (top-right to bottom-left)
        won = True
        for i in range(n):
            if board[i][n - 1 - i] != player:
                won = False
        if won:
            return player

    # Check if there are any empty cells left
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1  # game is still going

    return 0  # board is full and no winner = draw
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Exercise 5: Place a mark on the board at position (row, col)
# Returns True if the move was made, False if the cell was already taken
def place_mark(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    return False
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Exercise 7: Place a mark using a label like "A1" or "B3"
def place_mark_labeled(board, player, location):
    location = location.strip().upper()
    row_labels = ["A", "B", "C", "D", "E", "F", "G", "H"]
    n = len(board)

    # find which row index matches the letter
    row = -1
    for i in range(len(row_labels)):
        if location[0] == row_labels[i]:
            row = i

    # convert the column number (e.g. "2") to a 0-based index
    col = int(location[1:]) - 1

    # make sure the position is on the board
    if row == -1 or col < 0 or col >= n:
        return False

    return place_mark(board, player, row, col)
````

````python
# Test Exercise 7
b = create_board(3)
print(place_mark_labeled(b, 1, "A1"))  # True
print(place_mark_labeled(b, 2, "A1"))  # False - already taken
print(place_mark_labeled(b, 2, "B2"))  # True
draw_board_labeled(b)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
# Exercise 8: Ask a player to enter a valid location and place their mark
def player_turn(board, player):
    draw_board_labeled(board)
    if player == 1:
        symbol = "X"
    else:
        symbol = "O"

    while True:
        location = input("Player " + str(player) + " (" + symbol + "), enter location (e.g. A1): ")
        if place_mark_labeled(board, player, location):
            break
        print("That move is not valid. Try again.")
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Exercise 9: Full two-player tic-tac-toe game
def play_game(n=3):
    board = create_board(n)
    current_player = 1

    while True:
        player_turn(board, current_player)
        result = check_winner(board)

        if result == 1:
            draw_board_labeled(board)
            print("Player 1 (X) wins!")
            break
        elif result == 2:
            draw_board_labeled(board)
            print("Player 2 (O) wins!")
            break
        elif result == 0:
            draw_board_labeled(board)
            print("It's a draw!")
            break

        # switch to the other player
        if current_player == 1:
            current_player = 2
        else:
            current_player = 1
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Exercise 11: Simple computer player using basic strategy
import random

def computer_move_simple(board, computer_player):
    n = len(board)

    if computer_player == 1:
        human_player = 2
    else:
        human_player = 1

    # Step 1: check if computer can win in one move
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = computer_player
                if check_winner(board) == computer_player:
                    return  # winning move placed, done
                board[i][j] = 0  # undo

    # Step 2: block the human from winning
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = human_player
                if check_winner(board) == human_player:
                    board[i][j] = computer_player  # block
                    return
                board[i][j] = 0  # undo

    # Step 3: take the center if it is free
    mid = n // 2
    if board[mid][mid] == 0:
        board[mid][mid] = computer_player
        return

    # Step 4: pick any random empty cell
    empty_cells = []
    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                empty_cells.append((i, j))

    if len(empty_cells) > 0:
        choice = random.choice(empty_cells)
        board[choice[0]][choice[1]] = computer_player


def play_vs_computer(n=3):
    board = create_board(n)
    human = 1
    computer = 2
    current = 1

    while True:
        if current == human:
            player_turn(board, human)
        else:
            print("Computer is thinking...")
            computer_move_simple(board, computer)

        result = check_winner(board)

        if result == human:
            draw_board_labeled(board)
            print("You win!")
            break
        elif result == computer:
            draw_board_labeled(board)
            print("Computer wins!")
            break
        elif result == 0:
            draw_board_labeled(board)
            print("It's a draw!")
            break

        if current == 1:
            current = 2
        else:
            current = 1
````

````python
# Test Exercise 11
play_vs_computer(3)
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Exercise 12: Smarter computer player using minimax search
import math

def minimax(board, depth, is_maximizing, computer_player, human_player):
    result = check_winner(board)

    # base cases: someone won or board is full or depth limit reached
    if result == computer_player:
        return 10
    elif result == human_player:
        return -10
    elif result == 0 or depth == 0:
        return 0

    n = len(board)

    if is_maximizing:
        # computer's turn: look for the highest score
        best_score = -math.inf
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = computer_player       # try move
                    score = minimax(board, depth - 1, False, computer_player, human_player)
                    board[i][j] = 0                     # undo move
                    if score > best_score:
                        best_score = score
        return best_score

    else:
        # human's turn: look for the lowest score
        best_score = math.inf
        for i in range(n):
            for j in range(n):
                if board[i][j] == 0:
                    board[i][j] = human_player          # try move
                    score = minimax(board, depth - 1, True, computer_player, human_player)
                    board[i][j] = 0                     # undo move
                    if score < best_score:
                        best_score = score
        return best_score


def computer_move_minimax(board, computer_player, max_depth=4):
    if computer_player == 1:
        human_player = 2
    else:
        human_player = 1

    n = len(board)
    best_score = -math.inf
    best_row = -1
    best_col = -1

    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                board[i][j] = computer_player           # try move
                score = minimax(board, max_depth, False, computer_player, human_player)
                board[i][j] = 0                         # undo move
                if score > best_score:
                    best_score = score
                    best_row = i
                    best_col = j

    board[best_row][best_col] = computer_player


def play_vs_minimax(n=3, max_depth=4):
    board = create_board(n)
    human = 1
    computer = 2
    current = 1

    while True:
        if current == human:
            player_turn(board, human)
        else:
            print("Computer is thinking...")
            computer_move_minimax(board, computer, max_depth)

        result = check_winner(board)

        if result == human:
            draw_board_labeled(board)
            print("You win!")
            break
        elif result == computer:
            draw_board_labeled(board)
            print("Computer wins!")
            break
        elif result == 0:
            draw_board_labeled(board)
            print("It's a draw!")
            break

        if current == 1:
            current = 2
        else:
            current = 1
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Exercise 13: Have the two computer players play each other
def auto_game(n, p1_depth, p2_depth):
    board = create_board(n)
    current = 1

    while True:
        if current == 1:
            computer_move_minimax(board, 1, p1_depth)
        else:
            computer_move_simple(board, 2)

        result = check_winner(board)
        if result != -1:
            return result

        if current == 1:
            current = 2
        else:
            current = 1


def run_tournament():
    for n in [3, 4, 5]:
        p1_wins = 0
        p2_wins = 0
        draws = 0

        # use a smaller depth for bigger boards so it doesn't take too long
        if n == 3:
            depth = 4
        elif n == 4:
            depth = 3
        else:
            depth = 2

        for i in range(10):
            result = auto_game(n, p1_depth=depth, p2_depth=0)
            if result == 1:
                p1_wins = p1_wins + 1
            elif result == 2:
                p2_wins = p2_wins + 1
            else:
                draws = draws + 1

        print(str(n) + "x" + str(n) + " board - 10 games:")
        print("  Minimax (Player 1) wins: " + str(p1_wins))
        print("  Simple  (Player 2) wins: " + str(p2_wins))
        print("  Draws: " + str(draws))
        print("  Minimax win rate: " + str(p1_wins * 10) + "%")
        print("")

run_tournament()
````
